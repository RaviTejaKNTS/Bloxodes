## tl;dr
Wiki pages are the strongest non-catalog investment: they combine meaningful traffic scale with the highest robust engagement rate. Codes lead non-catalog traffic volume but are consumed quickly. Articles show promising reader depth but remain a small, concentrated portfolio and should be expanded selectively rather than broadly.

## Context & Methods

### Key Assumptions
- Source: self-hosted Umami PostgreSQL for `Bloxodes | bloxodes.com`.
- Window: July 12-21, 2026, ten complete Asia/Kolkata calendar days.
- Meaningful engagement: at least 30 active seconds and 50% page scroll.
- July 16 is excluded from engagement-rate comparisons because a concentrated Singapore laptop/Chrome spike produced 1,492 engaged events from 339 sessions, versus 2-19 such events on surrounding days. Traffic and momentum use all ten complete days.
- The source SQL is `2026-07-22-content-type-performance.sql`.

## Data

```python
import csv
from pathlib import Path

data_path = Path('2026-07-22-content-type-performance-data.csv')
with data_path.open(newline='', encoding='utf-8') as handle:
    rows = list(csv.DictReader(handle))

numeric_fields = [field for field in rows[0] if field != 'content_type']
for row in rows:
    for field in numeric_fields:
        row[field] = float(row[field])

len(rows), rows[0]
```

## Results

```python
non_catalog = [row for row in rows if row['content_type'] != 'catalog']
non_catalog_pageviews = sum(row['pageviews'] for row in non_catalog)
for row in non_catalog:
    row['non_catalog_share_pct'] = 100 * row['pageviews'] / non_catalog_pageviews

ranking = sorted(non_catalog, key=lambda row: row['pageviews'], reverse=True)
[(row['content_type'], round(row['non_catalog_share_pct'], 1), row['adjusted_engaged_rate_pct'], row['momentum_pct']) for row in ranking]
```

```python
articles = next(row for row in rows if row['content_type'] == 'article')
wikis = next(row for row in rows if row['content_type'] == 'wiki')
codes = next(row for row in rows if row['content_type'] == 'codes')
{
    'article_non_catalog_share_pct': round(100 * articles['pageviews'] / non_catalog_pageviews, 1),
    'article_adjusted_engagement_pct': articles['adjusted_engaged_rate_pct'],
    'wiki_adjusted_engagement_pct': wikis['adjusted_engaged_rate_pct'],
    'codes_non_catalog_share_pct': round(100 * codes['pageviews'] / non_catalog_pageviews, 1),
}
```

## Takeaways
1. Prioritize wiki hubs and collection pages.
2. Maintain codes for acquisition and use them to route visitors into deeper wiki/article content.
3. Expand articles selectively around evergreen guides, tier lists, and high-engagement game topics; avoid scaling the whole article portfolio until traffic broadens beyond the current top pages.
4. Treat events and checklists as experiments: their momentum is positive, but the absolute sample is still small.
