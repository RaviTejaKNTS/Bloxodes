## Takeaways
1. Protect the music and decal pages; their rankings and CTR show product-market fit for evergreen ID lookup intent.
2. Replicate the page contract only where Bing demand is already proven. Hair IDs, face IDs, and gear IDs are the best existing-surface opportunities.
3. Add natural ID/code synonyms to titles, H1s, intros, and FAQs only when the dataset genuinely fulfills that lookup intent.
4. Treat schema, word count, and update labels as supporting hygiene, not the growth engine.
